# Harware Acceleration


chequeamos que pytorch tenga acceso a la GPU de Nvidia o al MPS de Apple de no ser asi volvemos a la CPU

```python
if torch.cuda.is_available():
    device="cuda"
    print("Using GPU")
elif torch.backends.mps.is_available():
    device="mps"
    print("Using MPS")
else:
    device="cpu"
    print("Using CPU")
```

Creamos un tensor en la GPU

```python
M=torch.tensor([[1.,2.,3.],[4.,5.,6.]])
M=M.to(device)
```

para saber en que device está un tensor podemos verlo

```python
M.device
```

Tambien podemos crear un tensor directamente en la GPU usando el argumento device

```python
M=torch.tensor([[1.,2.,3.],[4.,5.,6.]], device=device)
M.device
```

Una vez que el tensor este en la GPU podemos correr operaciones normalmente y ellas tomaran lugar en la GPU

```python
R=M@M.T
R
```

# Autograd
Su función principal es calcular los gradientes (derivadas parciales) de forma automática aplicando la regla de la cadena, lo que permite entrenar redes neuronales sin necesidad de calcular derivadas a mano.


```python
x=torch.tensor(5.0,requires_grad=True) #requires_grad=True es para decirle a pytorch que es una variable no una constante
f=x**2
f
```

```python
f.backward()
x.grad
```

### gradient descent step
para hacerlo necesitamos situarlo dentro de torch.no_grad()

```python
learning_rate=0.1
with torch.no_grad():
    x-=learning_rate*x.grad #gradient descent step
```

antes de que repitas el proceso es importante hacer zero los gradientes

```python
x.grad.zero_()
```

poniendo todo junto el bucle de entrenamineto luciria asi

```python
learning_rate=0.1
x=torch.tensor(8.0,requires_grad=True)
for i in range(100):
  f=x**2 #forward pass
  f.backward() #backward pass
  with torch.no_grad():
    x-=learning_rate*x.grad #gradiente descent step
  x.grad.zero_() #reset the gradient
```
