# Harware Acceleration


chequeamos que pytorch tenga acceso a la GPU de Nvidia o al MPS de Apple de no ser asi volvemos a la CPU

```python
if torch.cuda.is_available():
    device="cuda"
    print("Using GPU")
elif torch.backends.mps.is_available():
    device="mps"
    print("Using MPS")
else:
    device="cpu"
    print("Using CPU")
```

Creamos un tensor en la GPU

```python
M=torch.tensor([[1.,2.,3.],[4.,5.,6.]])
M=M.to(device)
```

para saber en que device está un tensor podemos verlo

```python
M.device
```

Tambien podemos crear un tensor directamente en la GPU usando el argumento device

```python
M=torch.tensor([[1.,2.,3.],[4.,5.,6.]], device=device)
M.device
```

Una vez que el tensor este en la GPU podemos correr operaciones normalmente y ellas tomaran lugar en la GPU

```python
R=M@M.T
R
```
